```python
# Student Info Sheet

print("Hi !! Please fill out this form before proceeding.")

#Ask user for input

firstname = (str(input("First Name: "))).title()
lastname = (str(input("Last Name: "))).title()
section = (str(input("Section: "))).title()
favoritesubject = (str(input("Favorite Subject: "))).title()

# Make Arrays for easier coding

name = [firstname, lastname]
student_info = [firstname, lastname, section, favoritesubject]
total_characters = len("".join(student_info))

# Start printing

print("\nFormatted Output:")

# Utilize Arrays and functions to join, capitalize, and count the characters
print("Name:", " ".join(name))
print("Name Code:", firstname.upper(), lastname.upper())
print("Section:", section.upper())
print("Favorite Subject:", favoritesubject)
print("Character Count:", (total_characters))

# End of Code
```
