```python
# Student Info Sheet

print("Hi !! Please fill out this form before proceeding.")
e = (str(input("First Name: "))).title()
a = (str(input("Last Name: "))).title()
b = (str(input("Section: "))).title()
c = (str(input("Favorite Subject: "))).title()
name = [e, a]
student_info = [e, a, b, c]
total_characters = len("".join(student_info))
print("\nFormatted Output:")
print("Name:", " ".join(name))
print("Name Code:", e.upper(), a.upper())
print("Section:", b.upper())
print("Favorite Subject:", c)
print("Character Count:", (total_characters))
```
